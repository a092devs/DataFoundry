```python
from pathlib import Path
import pandas as pd

RAW_DATA = Path("../data/raw")

print(RAW_DATA.resolve())
```

```python
csv_files = sorted(RAW_DATA.glob("*.csv"))

for file in csv_files:
    print(file.name)
```

```python
datasets = {
    file.stem: pd.read_csv(file)
    for file in csv_files
}

print(f"Loaded {len(datasets)} datasets")
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(orders[column])
```

```python
products = datasets["olist_products_dataset"]
missing_product_info = products[
    products["product_category_name"].isna()
]

print("Products with missing category:", len(missing_product_info))

missing_product_info.head()
```

```python
missing_product_info.isna().sum()
```

```python
order_items = datasets["olist_order_items_dataset"]

missing_product_ids = set(missing_product_info["product_id"])
ordered_product_ids = set(order_items["product_id"])

used_missing_products = missing_product_ids & ordered_product_ids

print("Missing-info products:", len(missing_product_ids))
print("Missing-info products used in orders:", len(used_missing_products))
```

```python
print("Orders without matching customers:",
      len(set(orders["customer_id"]) - set(customers["customer_id"])))

print("Order items without matching orders:",
      len(set(order_items["order_id"]) - set(orders["order_id"])))

products = datasets["olist_products_dataset"]
sellers = datasets["olist_sellers_dataset"]
payments = datasets["olist_order_payments_dataset"]
reviews = datasets["olist_order_reviews_dataset"]

print("Order items without matching products:",
      len(set(order_items["product_id"]) - set(products["product_id"])))

print("Order items without matching sellers:",
      len(set(order_items["seller_id"]) - set(sellers["seller_id"])))

print("Payments without matching orders:",
      len(set(payments["order_id"]) - set(orders["order_id"])))

print("Reviews without matching orders:",
      len(set(reviews["order_id"]) - set(orders["order_id"])))
```

```python
orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)
```
